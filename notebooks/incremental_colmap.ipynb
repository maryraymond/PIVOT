```python
dataset = Path("/workspace/datasets/processed/incremental_recon_tight_2")
colmap_prefix = "PYCOLMAP_sp_"

os.makedirs(dataset, exist_ok=True)
colmap_dir = dataset / f"{colmap_prefix}anchor"
```

```python
# copy the required trajectories
def copy_traj(traj_list, src_data, dataset):
    for traj in traj_list:
        #check if this is a file or a folder
        if ".MP4"  in traj:
            # copy file
            shutil.copyfile(f"{src_data}/{traj}", f"{str(dataset)}/{traj}")
        else:
            shutil.copytree(f"{src_data}/{traj}", f"{str(dataset)}/{traj}")
```

```python
# run for anchor 
copy_traj(anchor, src_data, dataset)
dataset_images_dir = dataset
colmap_dir = dataset / f"{colmap_prefix}anchor"
sparse_name = "anchor"
```
