```python
data_path = "/data/datasets/processed/village_street/trajectories"
```

```python
folders = [f.name for f in Path(data_path).iterdir() if f.is_dir()]
```

```python
print(folders)
```

```python
def _make_metadata_dispatcher(metadata_map):
    def dispatch(image_path, use_absolute_altitude):
        traj_name = Path(image_path).parent.name
        return metadata_map[traj_name](image_path, use_absolute_altitude)
    return dispatch
```

```python
supported_image_capture_devices = {"dji_drone_mini_4_pro": lambda name, abs_alt: ( DjiDroneMini4ImageMetaData(image_file=name, 
                                                                                                        camera_tags_map=DjiDroneMini4TagsMap(),
                                                                                                        gps_tags_map=GpsTagsMap(),
                                                                                                        absolute_altitude=abs_alt)
                                                                                  )
                                    }
```

```python
image_metadata_map = {}
for traj_name in folders:
    image_metadata_map[traj_name] = supported_image_capture_devices["dji_drone_mini_4_pro"]
```

```python
create_image_metadata=_make_metadata_dispatcher(image_metadata_map)
```

```python
images_sub_folders = read_images_data_from_sub_folders(data_path=data_path, 
                                      create_image_metadata_map=image_metadata_map)
```

```python
image_subfolder_manual = {}
```

```python
for image in images_folders:
    folder_name = str(Path(image['file_name']).parent)
    image['file_name'] = Path(image['file_name']).name
    image_subfolder_manual.setdefault(folder_name, []).append(image)
```
